```python
import pandas as pd 
```

```python
DATA_PATH=r"E:\car-price-prediction\data\bonbanh\cleaned\cleaned_bonbanh.csv"
```

```python
car_df = pd.read_csv(DATA_PATH)
```

```python
car_df.head(5)
```

```python
brands = car_df.brand.unique()
```

```python
from pathlib import Path

OUT_DIR = Path("brand_reports")
OUT_DIR.mkdir(exist_ok=True)

df = car_df.copy()

for brand, brand_df in df.groupby("brand"):
    safe_brand = str(brand).replace("/", "_").replace("\\", "_")
    out_path = OUT_DIR / f"{safe_brand}.txt"

    table = (
        brand_df
        .groupby(["brand", "model", "name"])
        .size()
        .reset_index(name="count")
        .sort_values(["model", "name"])
    )

    with open(out_path, "w", encoding="utf-8") as f:
        f.write(table.to_string(index=False))

print(f"Saved to: {OUT_DIR.resolve()}")
```
